```python
import matplotlib.pyplot as plt
import seaborn as sns

tips = sns.load_dataset("tips")

sns.barplot(
    data = tips,
    x = "day",
    y = "tip",
    hue="sex"
    
)
```

```python
tips = sns.load_dataset("tips")

sns.boxplot(
    data = tips,
    x = "day",
    y = "tip",
    hue = "sex"
)
```
