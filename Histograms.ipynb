```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(0)
scores = np.random.normal(70,10,100)

plt.axvline(
    x=45,
    label="Passing Marks",
    color="red",
    linewidth=2,
    linestyle="-",
)

plt.hist(scores,color="skyblue",edgecolor="black",bins=5)
```

```python
import matplotlib.pyplot as plt

# 1. Our simple example dataset (15 test scores)
scores = [55, 58, 62, 65, 68, 71, 74, 75, 77, 79, 82, 85, 88, 92, 95]

# 2. Method A: Pass a specific NUMBER of bins
# Python will automatically slice your data into 5 equal-sized buckets
plt.hist(scores, bins=5, edgecolor='black', color='skyblue')
plt.title("Histogram with 5 Equal Bins")
plt.xlabel("Scores")
plt.ylabel("Frequency")
plt.show()

# 3. Method B: Pass EXACT bin boundaries (Highly Recommended)
# This lets you control exactly where each bucket starts and stops
custom_bins = [50, 60, 70, 80, 90, 100]

plt.hist(scores, bins=custom_bins, edgecolor='black', color='lightgreen')
plt.title("Histogram with Custom Grade Bins (Width of 10)")
plt.xlabel("Grade Ranges")
plt.ylabel("Number of Students")
plt.show()
```
