```python
import matplotlib.pyplot as plt
import seaborn as sns

flights = sns.load_dataset("flights")

flights_pivot = flights.pivot(index="month",columns="year",values="passengers")
sns.heatmap(
    flights_pivot,
    cmap = "coolwarm",
    annot = True,
    fmt = "d"
)

plt.title("Passengers Per Year")
plt.tight_layout()
```
